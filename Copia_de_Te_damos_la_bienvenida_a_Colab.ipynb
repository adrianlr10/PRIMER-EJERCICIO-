```
print("=== Asistente Virtual ===")
print("Escribe 'salir' para terminar")

while True:

    mensaje_cliente = input("\nCliente: ")

    mensaje_cliente = mensaje_cliente.lower()

    if mensaje_cliente == "salir":
        print("Gracias por comunicarte. ¡Hasta luego!")
        break


    # Caso A: Compra
    if ("pedido" in mensaje_cliente or "comprar" in mensaje_cliente or "menu" in mensaje_cliente):

        if "vegana" in mensaje_cliente:
            print("Tenemos la Pizza Huerto por $12. ¿Deseas pedirla?")
        else:
            print("Aquí tienes nuestro menú clásico: Pepperoni, Hawaiana y Queso. ¿Cuál prefieres?")


    # Caso B: Envío
    elif ("estado" in mensaje_cliente or "donde" in mensaje_cliente or "demora" in mensaje_cliente):
        print("Tu pedido ya salió del horno y llegará en aproximadamente 15 minutos.")


    # Caso C: Queja
    elif ("queja" in mensaje_cliente or "fria" in mensaje_cliente or "mal" in mensaje_cliente):
        print("Lamentamos el inconveniente. Te transferimos con el administrador de turno.")


    # Caso D
    else:
        print("Aún estoy aprendiendo. Indica si deseas hacer un pedido, consultar un estado o presentar una queja.")
```
